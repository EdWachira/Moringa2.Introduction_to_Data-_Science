- Create a CSV called sales.csv:

Product,Quantity,Price
Laptop,5,80000
Mouse,20,1500
Keyboard,10,3000
Monitor,7,25000

- Print each rows
- read it as a list
- read it as a dictionery
- add two rows of data to the file
- Modify the code so that it prints only the product' names.
- Using the same file, print only the student records and not the header.


```python

import csv
sales = [
    ["laptop", 5,80000],
    ["mouse", 20,1500],
    ["keyboard", 10,3000],
    ["monitor", 7,25000]

]

with open("sales.csv", "w",newline = "") as file:
    writer = csv.writer(file) #read as a list
    writer. writerow(["product", "quantity", "price"])
    writer.writerows(sales)

with open("sales.csv", "r") as file2:
    #reading as a list
    reader = csv.reader(file2)
    print("\nreading as a list\n")
    for r in reader:
        print(r)


with open("sales.csv", "r") as file3:
    #reading as a dictionary
    reader3 = csv.DictReader(file3)
    print("\nreading as a dictionary\n")
    for r in reader3:
        print(r)

more_sale = ["watch2", 7, 5000]
more_sales2 = [
    ["car", 1, 1000000],
    ["phone", 4, 50000]
]

with open("sales.csv", "a",newline = "") as file4:
    writer = csv.writer(file4) #read as a list
    #writer. writerow(["product", "quantity", "price"])
    writer.writerow(more_sale)
    writer.writerows(more_sales2)


"""with open("sales.csv", "r") as file5:
    #reading as a list
    reader = csv.reader(file5)
    print("\nreading as a list5\n")
    for r in reader:
        print(r[0])"""

with open("sales.csv", "r") as file6:
    #reading as a list6
    reader = csv.reader(file6)
    print("\nreading the first column for products\n")
    for r in reader:
        print(r[0])

#print the student records without the header

with open("sales.csv", mode = "r", encoding= "utf-8") as file5:
    reader = csv.reader(file5)
    next(reader)
    print("\nskipping the  header\n")
    for read in reader:
        print(read)

```
