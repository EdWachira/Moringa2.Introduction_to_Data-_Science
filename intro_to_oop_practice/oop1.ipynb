### Class

A class is a blueprint for creating objects that share common properties and behavior<br>
A class is a template that defines objects, methods and attributes associated to an object.<br>

### Objects

An object is an instance of a specific class. Or an instance created from a specific class.<br>
Each object has its own unique set of data as defined by the class methods or attributes.<br>

### Method

A method is a function defined inside a class that operates on instances of that class.<br>

## INHERITANCE

It's a principle that allows one class (child) to inherit attributes and methods from another class(parent/ super).<br>
This means the sub_class inherits properties and behaviors of the super_class but can still define its own properties.<br>
Example:<br>
> we can have two bank clients, one is a regular client, the other is a premium client.<br>
> There's a section that deals with premium clients.<br>
If we were to program that we can have two classes, one for all clients and the other for premium clients. The latter will inherit from the super.<br>
All clients can have, name, account number etc<br>
A premium will have the above but also have their own properties like stock portfolio.<br>

```python
class BankClient:
    def __init__(self, email, name, account_number):
        self.name = name
        self.account_no = account_number
        self.email = email

class PremiumClient(BankClient):
    def __init__(self, name, email, account_number, stock_portfolio):
        super().__init__(name, account_number, email)
        self.stock = stock_portfolio
```
