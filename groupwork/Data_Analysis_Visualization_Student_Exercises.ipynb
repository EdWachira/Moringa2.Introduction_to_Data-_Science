# Data Cleaning Process

````python
copy_df = df.copy()
print(copy_df)
````

````python
# Checking duplicates
df.duplicated()
````

The dataset contains 5 duplicate rows. These duplicate records may lead to inaccurate analysis or cause certain students to be counted more than once. Therefore, the duplicate rows should be identified and removed to ensure the dataset contains unique records.

````python
# Remove duplicates
df = df.drop_duplicates()
````

````python
# verifying duplicate
df.duplicated()
````

The dataset initially contained 5 duplicate rows. These duplicate records were removed during the data cleaning process. After removing, the dataset have no duplicate rows, ensuring that the data was clean and ready for further analysis.

````python
# Checking the missing values
df.isna()
````

````python
# Checking the missing values
df.isna().sum()
````

The result above revealed that there are 29 missing values in:(Internet_Acces = 4, Study_Hours_Per_Week = 6, Attendance_Percentage = 5, Previous_GPA = 5, Quiz_Score = 5, Participation_Score = 4) while the remaining 13 columns has no missing values.

````python
numeric_cols = [
    "Study_Hours_Per_Week",
    "Attendance_Percentage",
    "Previous_GPA",
    "Quiz_Score",
    "Participation_Score"
]

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

  # Fill Internet_Access missing values with the mode
df["Internet_Access"] = df["Internet_Access"].fillna(
    df["Internet_Access"].mode()[0]

)

    
````

````python
# Check whether the missing values are gone
df.isna().sum()
````

After fixing the missing values, the dataset has been successfully cleaned. Numerical missing values were replaced using the median, while the categorical Internet_Access missing values were replaced using the mode. The dataset now contains no missing values, making it ready for further data analysis.

````python
# Calculate the average score
df["Overall_Score"].mean()
````

The average overall score of the students was 73.56. This indicates that, on average, students achieved an overall score around 73.56, providing a general measure of their academic performance.

````python
# Find the highest score
df["Overall_Score"].max()
````

The highest overall score was 91.1, while the lowest was 42.4. This shows that students had different levels of academic performance.

````python
# Count students in each course
df["Program"].value_counts()
````

````python
# Calculate the average overall score for each program
df.groupby("Program")["Overall_Score"].mean()
````

````python
# Standardize inconsistent program names
df["Program"] = df["Program"].replace({
    "DataScience": "Data Science",
    "data science": "Data Science",
    "Software engineering": "Software Engineering"
})
````

````python
# Check the program names again
df["Program"].unique()
````

````python
# Calculate the average overall score for each program
df.groupby("Program")["Overall_Score"].mean()
````

````python
# Calculate average score by program
average_score_by_program = df.groupby("Program")["Overall_Score"].mean()

# Create the bar chart
average_score_by_program.plot(
    kind="bar",
    title="Average Overall Score by Program",
    xlabel="Program",
    ylabel="Average Overall Score"
)
````

````python
df["Overall_Score"].plot(
    kind="hist",
    bins=10,
    title="Distribution of Overall Scores",
    xlabel="Overall Score",
    ylabel="Number of Students"
)
````

````python
# Number of students by programs
student_count_by_program = df["Program"].value_counts()
student_count_by_program
````

````python
# Create the bar chart for number of students by programs
student_count_by_program.plot(
    kind="bar",
    title="Number of Students by Program",
    xlabel="Program",
    ylabel="Number of Students"
)
````

````python
student_count_by_program = df["Program"].value_counts()

ax = student_count_by_program.plot(
    kind="bar",
    title="Number of Students by Program",
    xlabel="Program",
    ylabel="Number of Students"
)

ax.bar_label(ax.containers[0])
````
