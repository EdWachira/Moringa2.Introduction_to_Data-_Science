Option A: Read as Lists (csv.reader)

```python
#Reading as rows

import csv

# Open the file safely using 'with'
with open('salaries.csv', mode='r', encoding='utf-8') as file:
    salaries = csv.reader(file)
    
    # Optional: Skip the header row if your CSV has column titles
    header = next(salaries) 
    print(f"{header}")
    
    # Loop through the remaining rows
    for row in salaries:
        print(row)  # 'row' is a list, e.g., ['John', '30', 'New York']
        print(row[0])  # Access the first column specifically
```

Option B: Read as Dictionaries (csv.DictReader)

```python
import csv

with open("salaries.csv", mode = "r", encoding = "utf-8") as file:
    salaries = csv.DictReader(file)

    for salary in salaries:

        # Access data using the column names as keys
        print(salary["work_year"], salary["experience_level"], salary["job_title"], salary["salary"], salary["salary_currency"], salary["salary_in_usd"])



        #access all columns
        #print(salary)
```

```python

#Writing to a csv file
#this will always overwrite the file

import csv


student =[
    ["Mary", 30, "Nakuru"],
    ["Job", 27, "Nairobi"],
    ["Bob", 34, "Meru"]
]

with open("student.csv", "w", newline = "") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "age", "town"]) #header
    writer.writerows(student)    #rows


```

Adding to the student file using Append

```python
student2 = ["Maxmillian", 33, "Marigat"]

with open("student.csv", mode = "a", newline = "") as f:
    writer = csv.writer(f)
    writer.writerow(student2)  #to only add a single row use writerow, formultiples rows use writerows
```
