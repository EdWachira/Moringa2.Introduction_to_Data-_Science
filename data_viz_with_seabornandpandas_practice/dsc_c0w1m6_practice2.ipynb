6. Create histograms:
- Create a histogram for each feature (sepal_length, sepal_width, petal_length, petal_width) using the `hist()` function.
- Set the number of bins to 20 using the `bins` parameter.
- Set the title of each plot using `set_title()` with an appropriate title for each feature.
- Label the x-axis with the feature name and the y-axis as 'Frequency'.
- Adjust the layout of the subplots using `plt.tight_layout()`.

```python
# Create histograms
fig, axs = plt.subplots(2, 2, figsize=(10, 8))
axs = axs.flatten()

for i, feature in enumerate(['sepal_length', 'sepal_width', 'petal_length', 'petal_width']):
    axs[i].hist(iris[feature], bins=20)
    axs[i].set_title(f'Histogram of {feature}')
    axs[i].set_xlabel(feature)
    axs[i].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Your code here

import matplotlib.pyplot as plt
import seaborn as sns

# Load the iris dataset
iris = sns.load_dataset('iris')

# Define the features we want to plot
features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

# Create a 2x2 grid of subplots
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(12, 10))

# Flatten the axes array for easy iteration
axes = axes.flatten()

# Loop through each feature and create a histogram
for i, feature in enumerate(features):
    # Plot the histogram on the corresponding subplot axis
    axes[i].hist(iris[feature], bins=20, edgecolor = "black", color='green')
    
    # Set the title and axis labels
    axes[i].set_title(f'Distribution of {feature.replace("_", " ").title()}')
    axes[i].set_xlabel(feature.replace('_', ' ').title())
    axes[i].set_ylabel('Frequency')

# Adjust layout to prevent overlapping
#plt.tight_layout()

# Display the plots
plt.show()
```
