## CREATE
Create a table in a database.
Once a table is created what follows is inserting the data in a table.

**Syntax:** CREATE TABLE table_name (Attr1 Type1, Attr2 Type2, ...., Attr_n Type_n);

For Example: Create a Table named students with columns name, student_id, course, email, age, gender, bmi. The keys are student_id it's a primary key. name, course are strings, email can be a string though it has its own datatype in databases, gender can be captured as string though it has data type in SQL which boolean.

```python
import sqlite3
import pandas as pd

# Create a database named moringa
moringa_db = sqlite3.connect("moringa.db")

# Establish a connection with the db
# conn = sqlite3.connect("moringa.db")
# conn1 = moringa_db.cursor()

# Create a table named students inside the database
create_tbl_query = """
    CREATE TABLE IF NOT EXISTS students (
        student_id INT PRIMARY KEY NOT NULL,
        name TEXT NOT NULL,
        course TEXT NOT NULL,
        email VARCHAR(50) UNIQUE NOT NULL,
        gender VARCHAR(10) NOT NULL,
        bmi REAL NOT NULL
    );
    """

moringa_db.execute(create_tbl_query)
```

```python
cur=moringa_db.cursor()
```

```python
pd.read_sql("SELECT * FROM students", moringa_db)
```

## READ
Retrieving data from a database. It has three clauses:</br></br>
**SELECT:** Gets the attributes required</br>
**FROM:** Refers to the table returning the attributes</br>
**WHERE:** It's a condition to get the attributes based on a specific criteria. It's not compulsory.</br>

**Syntax:** </br> READ attr1, attr2 FROM table_name WHERE condition </br> READ * FROM table_name WHERE condition

```python
# Read all the records from the table

# Read the email of the student with student_id = 3
```

```python
#Read all records from the table
q=(
    """
    SELECT * FROM students 

    """
   
)
cur.execute(q)
rows= cur.fetchall()
for row in rows:
    
    print(row)
```

## UPDATE
Modifying a record/row value inside a table.</br></br>

**Syntax:** UPDATE table_name SET attr1 = value1, attribute2 = value2 WHERE condition

```python
# Update Mutheu's course to data science and her bmi = 23.5
q=("""
UPDATE students 
SET course = "Data science", bmi=23.5
WHERE name= "Mutheu"
""")
cur.execute(q)
```

## DELETE
Erasing/Deleting a value inside a table.

**Syntax:** DELETE FROM table_name WHERE condition

```python
# Delete the last student
q=("""
DELETE FROM students
WHERE student_id =104


""")
cur.execute(q)
```
