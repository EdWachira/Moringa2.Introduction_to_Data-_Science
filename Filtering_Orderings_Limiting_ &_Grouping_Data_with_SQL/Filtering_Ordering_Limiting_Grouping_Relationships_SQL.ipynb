# FILTERING, ORDERING, LIMITING, GROUPING & SQL RELATIONSHIPS

## FILTERING, ORDERING, LIMITING & GROUPING

- **Filtering:** Returning/Retrieving data that has met a specific condition. For Example, you can return data of employees earning more than 50K.
- **Ordering:** Arranging data in a specific sequence. For Example, returning student names in alphabetical order; A-Z (ascending) & Z-A (descending)
- **Limiting:** Restricting the number of records/rows returned by a query result. For Example, returning the first five students from a students table.
- **Grouping:** Organizing records with similar attributes. For example, categorizing students based on a specific course (data science, software engineering, cyber-security)

## SQL RELATIONSHIPS
SQL table relationships describe how records in one table connect to records in another table. They help us organise information without unnecessarily repeating it.
For example, a school might store students in one table and courses in another. Relationships allow us to connect each student to the courses they take.
There are three main types of relationships:

- **One-to-One:** One record connects to at most one record in another table. For Example, one student has one student card
- **One-to-Many:** One record can connect to many records in another table. For Example, one department has many students
- **Many-to-Many:** Many records can connect to many records in another table. For Example, many students take many courses



### Primary Keys and Foreign Keys
A ***primary*** key uniquely identifies each row in a table. It cannot contain duplicate values or NULL. Consider this `departments` table:

| **department_id** | **department_name** |
| --- | --- |
| 1 | Data Science |
| 2 | Software Engineering |

In the table above, department_id is the primary key with each department having a unique ID.
A ***foreign key*** is a column—or set of columns that refers to a key in another table. It references the other table’s primary key. Consider this `students` table:

| **student_id** | **student_name** | **department_id** |
| --- | --- | --- |
| 101 | Mary | 1 |
| 102 | Brian | 1 |
| 103 | Amina | 2 |

In the table above, student_id is the `primary` key because it uniquely identifies each student and department_id is a `foreign` key because it connects the student to a department.</br>
An interpretation of one of the rows is, Mary’s department_id is 1 upon which looking up department 1 we find out that Mary belongs to Data Science.</br>
***NB:*** Foreign key values can repeat; Mary and Brian can belong to the same department.

### One-to-One
A ***one-to-one*** relationship means a record in one table can connect to at most one record in another table. For example, each student can have one student card and each student card belongs to one student. We can have a `students` table with a ***primary key (student_id)*** as below:
| **student_id** | **student_name** | **department_id** |
| --- | --- | --- |
| 101 | Mary | 1 |
| 102 | Brian | 1 |
| 103 | Amina | 2 |

Then a `students_card` table as below:

| card_id | student_id |
| --- | --- |
| 501 | 101 |
| 502 | 102 |
| 503 | 103 |

The keys are of the table are: `card_id`; The primary key and `student_id`; a foreign key which is a reference to the student who owns the card. For Example, card 501 references student 101 who is Mary. Each card points to one student ID and no student ID is allowed on a second card.

### One-to-Many
A ***one-to-many*** relationship means one record in a table can be associated with several records in another table. For example, one department can have many students. Each student belongs to one department. The foreign key goes in the table on the “many” side—in this case, `students` table.

`Department Table`
| **department_id** | **department_name** |
| --- | --- |
| 1 | Data Science |
| 2 | Software Engineering |

<br></br>
`Students Table`
| **student_id** | **student_name** | **department_id** |
| --- | --- | --- |
| 101 | Mary | 1 |
| 102 | Brian | 1 |
| 103 | Amina | 2 |

Referencing the tables above:
- Department 1 has students Mary and Brian.
- Department 2 has student Amina.
- Each student belongs to one department.

This is a one-to-many relationship because one department can have many students OR several students can belong to the same department.</br>
***NB:*** `department_id = 1` appears twice in `Students` table and that is allowed since a foreign key can repeat. The foreign key goes on the many side which is the `Students` table.

### Many-to-Many
A ***many-to-many*** relationship means several records in one table can connect to several records in another. For Example, suppose:
- Mary takes SQL and Python.
- Brian also takes SQL and Python.

This means one student can take many courses and one course can have many students. This is a many-to-many relationship. This relationship may need a third table to hold the connections; this table in known as a `linking` table. To demonstrate this connection we'll have students, courses and enrollments tables.

`Student Table`
| **student_id** | **student_name** | **department_id** |
| --- | --- | --- |
| 101 | Mary | 1 |
| 102 | Brian | 1 |
| 103 | Amina | 2 |

`Courses Table`
| course_id | course_name |
| --- | --- |
| 201 | SQL |
| 202 | Python |

`Enrollments Table` (Linking Table)
| student_id | course_id |
| --- | --- |
| 101 | 201 |
| 101 | 202 |
| 102 | 201 |
| 102 | 202 |

Each enrollment is a row containing two IDs (`student_id` and `course_id`). For Example, the first row is student 101 who is Mary taking up course 201 which is SQL, therefore this can be said that Mary takes SQL. Notice that both columns in `Enrollments` table are foreign keys:

- student_id references students.
- course_id references courses.

Together, they can form a unique pair known as a composite primary key which is unique. This means:

- Student 101 can appear again with a different course.
- Course 201 can appear again with a different student.
- But the exact pair (101, 201) cannot appear twice.

```python
import sqlite3
import pandas as pd
```

```python
# create database
conn = sqlite3.connect('school.db')
cursor = conn.cursor()
```

```python
# Tables
# Students{id, name, email}
# Enrollments{enr_id, std_id, couse_id, grade}
# Courses{course_id, course_name, lec_id}
# Lecturers{lec_name, lec_id, lec_email}
```

```python
# Create tables

# 1. Student Table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students(
    student_id INTEGER PRIMARY KEY,
    name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE
    )
    """)

# 2. Course
cursor.execute("""
CREATE TABLE IF NOT EXISTS courses(
    course_id INTEGER PRIMARY KEY,
    course_name TEXT NOT NULL,
    lecturer_id INTEGER,
    FOREIGN KEY(lecturer_id) REFERENCES lecturers(lecturer_id)
    )
    """)

# 3. Lecturers
cursor.execute("""
CREATE TABLE IF NOT EXISTS lecturers(
    lecturer_id INTEGER PRIMARY KEY,
    lecturer_name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE
    )
    """)

# 4. Enrollments
cursor.execute("""
CREATE TABLE IF NOT EXISTS enrollment(
    enrol_id INTEGER PRIMARY KEY,
    student_id INTEGER,
    course_id INTEGER,
    FOREIGN KEY (student_id) REFERENCES students(student_id),
    FOREIGN KEY(course_id) REFERENCES courses(course_id)
    )
    """)

#conn.commit()
```

### Joins

1 Inner join = select students who have courses enrolled

```python
q = """
SELECT students.name, courses.course_name
FROM students
INNER JOIN enrollment
    ON students.student_id = enrollment.student_id
INNER JOIN courses 
    ON enrollment.course_id = courses.course_id
"""

cursor.execute(q)
results = cursor.fetchall()

for row in results:
    print(row)
```

output all our students Plus their courses(either enrolled or not)

```python
# left join(students, enrollment, sourses)
# left join(courses, enrollment, students) /// right join(students, enrollment, courses)
# alliases is a nickname that you give your tables or columns for ease of referencing or output readability

q = """
SELECT s.name AS student_name, c.course_name
FROM students AS s
LEFT JOIN enrollment AS e
    ON s.student_id = e.student_id
LEFT JOIN courses AS c
    ON e.course_id = c.course_id;

"""
cursor.execute(q)

results = cursor.fetchall()

for row in results:
    print(row)
```

Find student, course, lecturer(&grade) in one query

```python
# full outer join
q = """
SELECT s.name, c.course_name, l.lecturer_name
FROM students AS s
INNER JOIN enrollment AS e
    ON s.student_id = e.student_id
INNER JOIN courses AS c
    ON c.course_id = e.course_id
INNER JOIN lecturers AS l
    ON c.lecturer_id = l.lecturer_id
"""
df = pd.read_sql_query(q, conn)
print(df)
```
