```python
import sqlite3
import pandas as pd
```

```python
# Create a database.
conn=sqlite3.connect('school.db')
cur=conn.cursor()
```

```python
#Create tables for our database
#Students(id,name,email),
#Enrollments(enr_id,std_id,course_id),
#Courses(course_id,course_name,lecturer_id),
#Lecturers(lectr_id,lectr_name,email)
```

```python
#Create tables
# 1. students table
cur.execute("""
CREATE TABLE IF NOT EXISTS students(
    student_id INTEGER PRIMARY KEY,
    name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE
    )
    """)

#2. lecturers table
cur.execute("""
CREATE TABLE IF NOT EXISTS lecturers(
    lecturer_id INTEGER PRIMARY KEY,
    lecturer_name VARCHAR(50) NOT NULL,
    email VARCHAR(100) UNIQUE
    )
    """)

#3. courses table
cur.execute("""
CREATE TABLE IF NOT EXISTS courses(
    course_id INTEGER PRIMARY KEY,
    course_name TEXT NOT NULL,
    lecturer_id INTEGER,
    FOREIGN KEY(lecturer_id)REFERENCES lecturers(lecturer_id)
    )
    """)


#4. enrollments table 
cur.execute("""
CREATE TABLE IF NOT EXISTS enrollment(
    enrol_id INTEGER PRIMARY KEY,
    student_id INTEGER,
    course_id INTEGER,
    FOREIGN KEY(student_id)REFERENCES students(students_id),
    FOREIGN KEY (course_id) REFERENCES courses(course_id)
    )
    """)

conn.commit()
```

#### Q1.)**Inner join- select students who have courses enrolled**

```python
q="""
SELECT *
FROM students
INNER JOIN enrollment
    ON students.student_id=enrollment.student_id
INNER JOIN courses
    ON enrollment.course_id=courses.course_id
"""

cur.execute(q)

results=cur.fetchall()

for row in results:
    print(row)

#This brings out everything but we only want student and course details.
```

```python
q="""
SELECT students.name AS student_name,courses.course_name
FROM students
INNER JOIN enrollment
    ON students.student_id=enrollment.student_id
INNER JOIN courses
    ON enrollment.course_id=courses.course_id
"""

cur.execute(q)

results=cur.fetchall()

for row in results:
    print(row)
```

#### Q2.)**Output all our students Plus their courses(either enrolled or not)**


```python
#We will use a left join with the students on the left side 
#Left join students,enrollment,courses
#Use aliases to try and simplify your work

q="""
SELECT s.name AS student_name,c.course_name
FROM students AS s
LEFT JOIN enrollment AS e
    ON s.student_id=e.student_id
LEFT JOIN courses AS c
    ON e.course_id=c.course_id;
"""

cur.execute(q)

results=cur.fetchall()

for row in results:
    print(row)
```

#### Q3.)**Find student, course, lecturer(& grade) in one query**

```python
#Full outer join
q=""" 
SELECT s.name,c.course_name,l.lecturer_name
FROM students AS s
INNER JOIN enrollment AS e
    ON s.student_id=e.student_id
INNER JOIN courses AS c
    ON c.course_id=e.course_id
INNER JOIN lecturers AS l
    ON c.lecturer_id=l.lecturer_id;
"""

cur.execute(q)

results=cur.fetchall()

for row in results:
    print(row)
```
