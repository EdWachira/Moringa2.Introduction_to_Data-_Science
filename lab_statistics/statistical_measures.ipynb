Mean

```python
scores = [30, 36, 60, 55, 80, 90, 40, 45]

total = sum(scores)
print(total)
count = len(scores)
print(count)

mean = total/count
print(mean)
```

Using numpy

```python
import numpy as np


scores = [30, 36, 60, 55, 80, 90, 40, 45]

mean2 = np.mean(scores)
print(mean2)

median = np.median(scores)
print(median)
```

mode

by importing statistics

```python
import statistics
scores2 = [40, 40, 35, 80, 90, 50, 50, 50]

mode = statistics.mode(scores2)
print(mode)
```

max and min values

```python
print(max(scores2))
print(min(scores2))
```

variance

```python


variance_value = np.var(scores)
print(variance_value)

#variance = sum(squared difference)/ (len(scores))
```

Standard Deviation

```python
std = np.std(scores)
print(std)
```

Quartiless
Dividing data to 4 parts

quartile ranges

```python
q1 = np.percentile(scores, 25)
q2 = np.percentile(scores, 50)
q3 = np.percentile(scores, 75)
q4 = np.percentile(scores, 100)


iqr = q3 - q1

print(q1)
print(q2)
print(q3)
print(q4)
print(iqr)
```
