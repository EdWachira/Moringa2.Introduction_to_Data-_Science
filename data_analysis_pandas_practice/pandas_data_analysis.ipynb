## Introduction to data analysis using pandas

What is Data Analysis?
It is the process of examining data to find patterns, answer questions and make decisions. For example a number becomes useful whhen we understand what it measures, a waiting time of 20 means nothing until we know what the unit is in minutes.

Data Science is the field that uses data, statistics, domain knowledge and programming to investigate questions and come up with relevant solutions. Data analysis thus becomes a part of this work.

There are four cmmon types of analysis:
- ***Descriptive analysis:*** It tells us what happened, for example how many patient visits were recorded per clinic.
- ***Diagnostic analysis:*** Tells us why it happened, for example busy patient arrival period might result to longer wait times.
- ***Predictie analysis:*** Tells us what might happen, for eaxmple how many patients might be expected next week.
- ***Prescriptive analysis*** Tells us what should be done or what action should be taken, for eaxmple which staffing plan could reduce patient waiting times.

**Analysis Workflow**
- State a Question
- Get/Obtain/Load the data.
- Inspecting the data: Checking for null values, duplicate values, outliers/extreme values, data types, data size(rows,columns), erroneus values
- Clean the data
- Analyze and Visualize
- Interpret and communicate the results



**Pandas**
- It's a python library for working with tabular data (rows and columns). It can read files, perform data cleaning, aggregation functioons and coombine tables. Its main structure is the ***Series()*** and ***DataFrame()***.

## Pandas Series and DataFrames
**Series**
This is a one dimensional labelled collection of values. Think of it as a single column together with its rows

```python
import pandas as pd #pd ia an alias

#Pandas Series
age = pd.Series([22, 34, 27, 44, 19], name = "age")
age
```

```python
temperature = pd.Series([21, 25, 28, 29, 31, 22, 24], index = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"], name = "Temperatures")
temperature
```

**DataFrames ** Its a two dimensional labelled table with rows and columns i.e Like an excel sheet which is a collection of multiple series(has multiple rows and multiple columns). For example data that has name , age and height

```python
students = {
    "name": ["John", "Mary", "Peter", "Fridah"],
    "age": [22, 34, 27, 28],
    "county": ["Nairobi", "Mombasa", "Kisumu", "Kirinyaga"],
}
print(students)

df = pd.DataFrame(students)
df
```
