# Data Cleaning with Pandas

** What is data cleaning?** <br> Data cleaning is the process of identifying and addressing problems thhat make data unreliable/ difficult to analyze.

Common data cleaning problems include:
- Missing values
- Duplicate values
- Inconsistent text
- Incorrect Data Typing
- Invalid values
- An outlier - which is an unusually high or low value compared to the observation. It is not automatically an error. For example, if the patient wait time is normally 110 - 15 mins but there is a recorded value of a 90mins wait time. This is clearly an outlier but that does not mean it iss erroneous. We will end up deciding whether to drop or retain the value/outlier. However a negative wait time violates the column denoting that it is erroneous.

**Data Cleaning Processs**
- Preserve the raw data and create a working copy
- Inspect the data cleaning issues: Missing values...
- Deal with the issues example dropping null values, imputations, introduce unifrmity e.t.c
- Save the cleaned copy


```python
# Step 1: Looad the data
import pandas as pd

#clinic_visits = pd.read_csv("C:\\Users\\Wachira\\Desktop\\Files\\Studying\\moringa_school\\2.Introduction_to_Data _Science\\clinic_visits_raw.csv")
clinic_visits = pd.read_csv("clinic_visits_raw.csv")
print(clinic_visits)
#clinic_visits.head()

```

```python
#step 2: Make a copy of the raw data

clinic_visits_copy = clinic_visits.copy()
clinic_visits_copy.head(6)
```

Data Cleaning Steps
1. Check for missing values
2. Drop duplicates and Drop/Impute missing values
3. Check and handle outliers
4. Check data types
5. Standardize column names

```python
#Step 3: Inspect for null/missing and duplicate values

#print(clinic_visits_copy.isnull().sum())

print("\n \n")

print("\nchecking for null values\n")
print(clinic_visits_copy.isna().sum())

#Step 4 Check for Data Types
print("\nchecking Data types\n")

print(clinic_visits_copy.dtypes)

#Inspect for duplicate values
print("\nchecking for Duplicate rows\n")
print(clinic_visits_copy.duplicated().sum())

#identify the duplicated row.
print("\nidentifying the Duplicate row\n")
duplicate_rows = clinic_visits_copy[clinic_visits_copy.duplicated(keep=False)]
print(duplicate_rows)


#Remove the exact duplicates

print("\nRemoving the Duplicate row\n")
clinic_visits_copy = clinic_visits_copy.drop_duplicates().reset_index(drop=True)

clinic_visits_copy

#Standardize the columns(Remove the white spaces and consistent naming)

clinic_visits_copy.columns = clinic_visits_copy.columns.str.strip().str.lower()

print("\nStandardize the columns \n")
clinic_visits_copy

#Convert data types for the following: fees(int) and date(datetime)

clinic_visits_copy["visit_date"] = pd.to_datetime(
    clinic_visits_copy["visit_date"]
)
#check for invalid ages and missing times
#choose between retaining, dropping and filling(imputing)
#validate the cleaned data and export(save) it as a CSV file



```
