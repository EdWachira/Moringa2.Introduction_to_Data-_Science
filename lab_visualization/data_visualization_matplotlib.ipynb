```python
#Example of a Pie chart using matplotlib

import matplotlib.pyplot as plt

# 1. Define your data
categories = ['Apples', 'Bananas', 'Cherries', 'Dates']
shares = [35, 25, 25, 15]

# 2. Configure aesthetic options
custom_colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
explode_slice = (0.02, 0.02, 0.02, 0.02)  # Explode/pop out the 1st slice (Apples) by 2%

# 3. Create the pie chart
plt.figure(figsize=(5, 5))  # Keeps the chart perfectly square
plt.pie(
    shares, 
    explode=explode_slice, 
    labels=categories, 
    colors=custom_colors,
    autopct='%1.1f%%',       # Formats the inside numbers as percentages
    startangle=140,          # Rotates the starting point of the chart
    shadow=True              # Adds a subtle drop shadow effect
)

# 4. Final touches
plt.title('Fruit Market Share Allocation')
plt.show()
```

Suppose we have data on the favorite fruits of a group of people. Let's create a bar chart to visualize this qualitative data.

```python
import matplotlib.pyplot as plt

# Data
fruits = ['Apple', 'Banana', 'Orange', 'Grape']
counts = [25, 20, 15, 10]

# Create a bar chart
plt.figure(figsize=(6, 4))
plt.bar(fruits, counts)

# Add labels and title
plt.xlabel('Fruit')
plt.ylabel('Count')
plt.title('Favorite Fruits')

# Display the chart
plt.show()
```

Explanation:

We import the matplotlib.pyplot module and alias it as plt.
We define the fruits list containing the categories and the counts list with the corresponding counts.
We create a new figure with a specified size using plt.figure(figsize=(6, 4)).
We create a bar chart using plt.bar(), passing the fruits and counts as arguments.
We add labels for the x-axis and y-axis using plt.xlabel() and plt.ylabel(), respectively.
We set a title for the chart using plt.title().
Finally, we display the chart using plt.show().
